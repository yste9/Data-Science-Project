### Counting 'voiture' in a list of items

If you have a list of strings and want to count how many times 'voiture' appears (case-sensitive), you can use a loop or the `count()` method.

```python
items = ['voiture', 'camion', 'moto', 'voiture', 'velo', 'voiture', 'avion']

# Method 1: Using a loop
count_loop = 0
for item in items:
    if item == 'voiture':
        count_loop += 1
print(f"Number of 'voiture' (using loop): {count_loop}")

# Method 2: Using the list.count() method
count_method = items.count('voiture')
print(f"Number of 'voiture' (using .count()): {count_method}")
```

### Counting 'voiture' in a text string

If you have a block of text and want to count the occurrences of the word 'voiture', you can use string methods, often combined with `split()` for word-based counting, or regular expressions for more complex pattern matching.

```python
text = "J'ai vu une voiture rouge et une autre voiture bleue. La voiture était garée devant la maison."

# Count case-insensitively and handle punctuation for a simple word count
text_lower = text.lower()
words = text_lower.replace('.', '').replace(',', '').split()

count_in_text = words.count('voiture')
print(f"Number of 'voiture' in text: {count_in_text}")
```
